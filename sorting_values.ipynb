# Sorting in DataFrame

```
import pandas as pd
```

```
df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Score": [60, 20, 90, 87]
})
```

```
# Sort by score ascending
df_sort_ascend = df.sort_values(by="Score", ascending=True)
df_sort_ascend
```

```
# Sort by score descending
df_sort_descend = df.sort_values(by="Score", ascending=False)
df_sort_descend
```

```
# Sort by Multiple columns
df.sort_values(by=["Score", "Name"], ascending=[True, False])
```

Sort DataFrame by index

```
df.sort_index(ascending=[True, False])
```

Sort Series

```
s = pd.Series(["Alice", "Bob", "Charlie", "David"])
sorted_series = s.sort_values(ascending=False)
print(sorted_series)

# Using the in
```

### Using the inplace=True directly modifies the series or dataframe without returning new object

```
# Sorting by key functions
sort_key_func = df.sort_values(by="Score", ascending=True, key=lambda x: x.astype(float))

print(sort_key_func)
```
