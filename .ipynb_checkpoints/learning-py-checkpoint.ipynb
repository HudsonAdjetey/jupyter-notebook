**Creating dataframe from an array**

```python
import pandas as pd
import numpy as np
```

```python
# Creating an array
arr = np.array([[1,4], [3, 6], [3, 6]])
print(arr * 2)
```

```python
# Creating a dataframe
df = pd.DataFrame(arr, index=['row1', 'row2', 'row3'], columns=['col1', 'col2'])
```

```python
df
```

***Option 2***

```python
data = [[1,4], [3, 6], [3, 6]]

df = pd.DataFrame(data, columns=['cols1', 'cols2'], index=["rol1", "rol2", "rol3"])
```

```python
df
```

***Creating a Dataframe from a dictionary***

```python
# List used for this example
states = ["California", "Texas", "Florida", "New York"]
population = [39903, 232323, 31113, 90009]
```

```python
# Storing lists within a dictionary
dict_states = {"states": states, "Population": population}
```

```python
print(dict_states)
```

```python
# Creating the dataframe
df_population = pd.DataFrame(dict_states)
```

```python
df_population
```
