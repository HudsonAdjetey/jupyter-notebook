***Creating a series using the random numbers***

```python
score1 = np.random.randint(1, 100, size=1000)
score2 = np.random.randint(1, 100, size=1000)
```

***Creating series using the random numbers***

```python
series1 = pd.Series(score1, index=np.arange(0, 1000))
series2 = pd.Series(score2, index=np.arange(0, 1000))
```

```python
pd.describe_option()
```

```python
series1.describe()
```

```python
df_new = pd.DataFrame({})
```

```python
df_new = df_new.assign(score1=series1, score2=series2)
```

```python
df_new
```

Math Operations

```python
# Select a column and calculate total sum
total = df_new["score1"].sum() + df_new["score2"].sum()
print(total)
```

# Count, mean, std, max and min

```python
# Count, mean, std, max and min
total = df_new[["score1" ,"score2"]].count()
# print(total)
mean = df_new["score1"].mean()
# print(mean)
std = df_new["score1"].std()
# print(std)
max_value = df_new["score2"].max()
# print(max_value)
min_value = df_new["score2"].min()
# print(min_value)
```

# Operating in rows

```python
# Calculating the sum in a row
df_new["Total"] =  df_new.sum(axis=1)
df_new
```

```python
df_new = df_new.round(2)
df_new
```

## Value counts

```python
# Counting element
# len function
len(df_new["score2"])
# Using the count
df_new["score2"].count()

```

```python
df_new["score2"].value_counts()
```

```python
# Return the relative frequency (divide all values by sum of the values)
df_new["score2"].value_counts(normalize=True)
```

```python
# converting the frequency to percentage
df_new["score1"].value_counts(normalize=True) * 100

for col in df_new.columns:
    print(f'Relative frequency for {col}')
    print(df_new[col].value_counts(normalize=True))
```

```python
# Counting by category
seriesCategory = pd.Series(np.random.randint(0, 100, size=1000))
df_new['Category'] = seriesCategory

counts = df_new["Category"].value_counts()
counts_cat = df_new.groupby("score1")["Category"].size()
counts_cat
```
